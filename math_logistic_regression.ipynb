```python
# sigmoid
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
def data_preparation(X):
  #initialize theta with 0 (and the same number of feature i mean the number of feature is the same as the theta also each feature have his theta)
  n_features = X.shape[1]
  #the +1 is for theta zero (the bias term)
  theta = np.zeros(n_features + 1)
  # Add a column of 1s at position 0 for θ₀.
  X_with_bias = np.c_[np.ones(X.shape[0]), X]

  return X_with_bias, theta

def proba_predict(X, theta):
    z = np.dot(X, theta)
    return sigmoid(z)
```

 Cost Function
J(θ)==>
Measures how wrong the predictions are (penalizes bad predictions).
image.png


```python
def cost_function(X,Y,theta):
  n=len(Y)
  h=proba_predict(X,theta)
  cost=-(1/n) *np.sum(Y*np.log(h)+(1-Y)*np.log(1-h))
  return cost

```

#### radient descent
update theta

image.png

```python
def gradient_descent(X,Y,theta,alpha,num_iterations):
  n=len(Y)
  cost_path=[]

  for _ in range(num_iterations):
    h=proba_predict(X,theta)
    #∂J(θ)/∂θj
    gradient=np.dot(X.T,(h-Y))
    theta -= alpha * gradient
    cost=cost_function(X,Y,theta)
    cost_path.append(cost)
  return theta,cost_path
```

## train the model


```python
alpha=0.01 #alpha is the learning rate
iteration_number=1000
X_with_theta_bias,theta = data_preparation(X)
theta_result, costs= gradient_descent(X_with_theta_bias,Y,theta,alpha,iteration_number)
```

# prediction

```python
# testing data
#df_predict= pd.read_csv('/content/Copy of sonar data.csv',header=None)
```

```python
def predict(new_X,theta):
  #add bias term
  new_X_with_bias = np.c_[np.ones(new_X.shape[0]), new_X]
  probabilities =proba_predict(new_X_with_bias,theta)
  return probabilities , (probabilities >= 0.5).astype(int)  # Convert to 0/1 predictions


```

```python
proba,y_predicted =predict(X,theta_result)
y_predicted
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

# Accuracy
accuracy = accuracy_score(Y, y_predicted)
print(f"Accuracy: {accuracy:.2f}")

# Confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(Y, y_predicted))
```
